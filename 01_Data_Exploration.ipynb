```python
# Imports and paths
from pathlib import Path
import pandas as pd

DATA = Path(r"C:\AI Projects\PopulationHealthWorkbench\data\raw")

```

Load all CSV files

```python
tables = {}

for csv_file in DATA.glob("*.csv"):
    table_name = csv_file.stem
    tables[table_name] = pd.read_csv(csv_file)

print(f"Loaded {len(tables)} tables.")
```

Data Inventory

```python
summary = []

for name, df in tables.items():
    summary.append({
        "Table": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Memory (KB)": round(df.memory_usage(deep=True).sum()/1024,2)
    })

inventory = pd.DataFrame(summary)

inventory.sort_values("Rows", ascending=False)
```

Data Dictionary

```python
dictionary = []

for table_name, df in tables.items():

    for col in df.columns:

        dictionary.append({
            "Table": table_name,
            "Column": col,
            "Data Type": str(df[col].dtype),
            "Missing Values": int(df[col].isna().sum()),
            "Unique Values": int(df[col].nunique())
        })

data_dictionary = pd.DataFrame(dictionary)
```
